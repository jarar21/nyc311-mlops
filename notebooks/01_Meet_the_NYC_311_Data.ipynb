# Stage 1 — Understand the project

**What:** We will predict whether a new NYC 311 request is still open 72 hours later.

**Why:** Every later choice—data, model, testing, monitoring—depends on a clear question.

One **row** is one request. A **feature** is information known when it arrives. A **label** is the yes/no answer learned later.

We will use [NYC Open Data](https://data.cityofnewyork.us/Social-Services/311-Service-Requests-from-2020-to-Present/erm2-nwe9).

```python
# Cell 2 — Why: see real rows before we build anything.
import pandas as pd
import requests

url = "https://data.cityofnewyork.us/resource/erm2-nwe9.json"
fields = ["unique_key", "created_date", "closed_date", "complaint_type", "borough"]
params = {"$select": ",".join(fields), "$order": "created_date DESC", "$limit": 10}

reply = requests.get(url, params=params, timeout=30)
reply.raise_for_status()
sample = pd.DataFrame(reply.json()).reindex(columns=fields)
display(sample)
```

```python
# Cell 3 — Why: check whether this small sample has duplicates or blanks.
print("Rows, columns:", sample.shape)
print("IDs unique:", sample["unique_key"].is_unique)
display(sample.isna().sum().to_frame("blank_values"))
```

# Stage 1 takeaway

**Why these cells mattered:** We confirmed the data source works and learned what the columns mean before training.

- `created_date`, `complaint_type`, and `borough` are possible inputs.
- `unique_key` identifies a request; it is not a useful prediction input.
- `closed_date` helps create the answer later. Giving it to the model would reveal the answer.
- A blank `closed_date` is not automatically a late request. If the request is only 2 hours old, its 72-hour label is still **unknown**.
